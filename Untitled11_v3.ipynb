## 4. Preprocessing

```python
df_clean = df.drop(['RowNumber', 'CustomerId', 'Surname'], axis=1)
X = df_clean.drop('Exited', axis=1)
y = df_clean['Exited'].values

# One-hot encode categoricals
cat_cols = X.select_dtypes(include='object').columns
X = pd.get_dummies(X, columns=cat_cols, drop_first=True)
feature_names = X.columns.tolist()

print(f'Features: {X.shape[1]}  |  Samples: {X.shape[0]}')
print(f'Feature names: {feature_names}')
X.head()
```

```python
# Stratified 80/20 split
X_train_raw, X_test_raw, y_train, y_test = train_test_split(
    X.values, y, test_size=0.2, random_state=SEED, stratify=y
)

# Scale BEFORE SMOTE — avoids leakage
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train_raw)
X_test_scaled  = scaler.transform(X_test_raw)

print(f'Train: {X_train_scaled.shape}  Churn: {y_train.mean():.2%}')
print(f'Test:  {X_test_scaled.shape}   Churn: {y_test.mean():.2%}')
```

```python
# SMOTE — only on TRAINING set to prevent data leakage
smote = SMOTE(random_state=SEED)
X_train_sm, y_train_sm = smote.fit_resample(X_train_scaled, y_train)

print(f'Before SMOTE: {X_train_scaled.shape[0]} samples  | Churn: {y_train.mean():.2%}')
print(f'After  SMOTE: {X_train_sm.shape[0]} samples  | Churn: {y_train_sm.mean():.2%}')

fig, axes = plt.subplots(1, 2, figsize=(10, 4))
for ax, y_data, title in zip(axes, [y_train, y_train_sm], ['Before SMOTE', 'After SMOTE']):
    counts = pd.Series(y_data).value_counts()
    ax.bar(['Retained','Churned'], counts.sort_index(), color=['#2ecc71','#e74c3c'], edgecolor='white')
    ax.set_title(title, fontweight='bold'); ax.set_ylabel('Count')
    for i, v in enumerate(counts.sort_index()):
        ax.text(i, v+100, str(v), ha='center', fontsize=11)
plt.suptitle('Class Distribution: Before vs After SMOTE', fontsize=13, fontweight='bold')
plt.tight_layout(); plt.show()
```

## 5. Helper Functions

```python
def evaluate_model(model_name, y_true, y_pred, y_prob):
    """Return a dict of all key metrics."""
    return {
        'Model':     model_name,
        'Accuracy':  round((y_true == y_pred).mean(), 4),
        'AUC-ROC':   round(roc_auc_score(y_true, y_prob), 4),
        'Avg Prec':  round(average_precision_score(y_true, y_prob), 4),
        'Precision': round(precision_score(y_true, y_pred, zero_division=0), 4),
        'Recall':    round(recall_score(y_true, y_pred), 4),
        'F1-Score':  round(f1_score(y_true, y_pred), 4),
    }

def find_best_threshold(y_true, y_prob, metric='f1'):
    """Find optimal classification threshold."""
    thresholds = np.arange(0.05, 0.95, 0.01)
    if metric == 'f1':
        scores = [f1_score(y_true, (y_prob>=t).astype(int), pos_label=1, zero_division=0) for t in thresholds]
    else:
        scores = [recall_score(y_true, (y_prob>=t).astype(int), zero_division=0) for t in thresholds]
    best_t = thresholds[np.argmax(scores)]
    return best_t, max(scores)

def plot_confusion_matrix(ax, y_true, y_pred, title):
    """Plot annotated confusion matrix on given axes."""
    cm = confusion_matrix(y_true, y_pred)
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=ax,
                xticklabels=['Retained','Churned'],
                yticklabels=['Retained','Churned'],
                linewidths=1, linecolor='white')
    ax.set_title(title, fontsize=11, fontweight='bold')
    ax.set_xlabel('Predicted'); ax.set_ylabel('Actual')
    total = cm.sum()
    for i in range(2):
        for j in range(2):
            ax.text(j+0.5, i+0.72, f'({cm[i,j]/total:.1%})',
                    ha='center', fontsize=8, color='gray')

all_results = []  # Collects metrics for all models
print('Helpers defined.')
```

## 6. Baseline ML Models (for Comparison)

> LogisticRegression, RandomForest, and XGBoost are trained on SMOTE data
> to establish a performance ceiling that the ANN should beat.

```python
baseline_models = {
    'LogisticRegression': LogisticRegression(max_iter=1000, random_state=SEED, class_weight='balanced'),
    'RandomForest':       RandomForestClassifier(n_estimators=200, random_state=SEED, class_weight='balanced', n_jobs=-1),
    'XGBoost':            XGBClassifier(n_estimators=200, scale_pos_weight=4, random_state=SEED,
                                         use_label_encoder=False, eval_metric='logloss', verbosity=0)
}

print('Training baseline models...')
for name, clf in baseline_models.items():
    clf.fit(X_train_sm, y_train_sm)
    y_prob = clf.predict_proba(X_test_scaled)[:, 1]
    best_t, _ = find_best_threshold(y_test, y_prob)
    y_pred = (y_prob >= best_t).astype(int)
    metrics = evaluate_model(name, y_test, y_pred, y_prob)
    all_results.append(metrics)
    print(f'{name:25s}  AUC={metrics["AUC-ROC"]:.4f}  F1={metrics["F1-Score"]:.4f}  Recall={metrics["Recall"]:.4f}')

print('Baseline training complete.')
```

## 7. Hyperparameter Search with Keras Tuner (Hyperband)

> Searches over: number of units per layer, dropout rate, learning rate, and number of layers.
> Hyperband is efficient — prunes bad trials early.

```python
def build_model_for_tuner(hp):
    """Model builder function for Keras Tuner."""
    model = keras.Sequential()

    # Tune number of hidden layers: 2 or 3
    n_layers = hp.Int('n_layers', min_value=2, max_value=3)

    # Tune units and dropout for each layer
    for i in range(n_layers):
        units = hp.Int(f'units_{i}', min_value=32, max_value=256, step=32)
        model.add(layers.Dense(units, activation='relu',
                               input_shape=(X_train_sm.shape[1],) if i==0 else ()))
        model.add(layers.BatchNormalization())
        dropout = hp.Float(f'dropout_{i}', min_value=0.1, max_value=0.5, step=0.1)
        model.add(layers.Dropout(dropout))

    model.add(layers.Dense(1, activation='sigmoid'))

    # Tune learning rate
    lr = hp.Float('learning_rate', min_value=1e-4, max_value=1e-2, sampling='log')

    model.compile(
        optimizer=keras.optimizers.Adam(learning_rate=lr),
        loss='binary_crossentropy',
        metrics=['accuracy', AUC(name='auc'), Recall(name='recall')]
    )
    return model

print('Tuner builder function defined.')
```

```python
# Hyperband tuner — fast, efficient
tuner = kt.Hyperband(
    build_model_for_tuner,
    objective=kt.Objective('val_auc', direction='max'),
    max_epochs=50,
    factor=3,
    directory='kt_bank_churn',
    project_name='ann_hyperband',
    overwrite=True,
    seed=SEED
)

print('Keras Tuner search space summary:')
tuner.search_space_summary()
```

```python
# Run hyperparameter search
early_stop_tuner = EarlyStopping(monitor='val_auc', patience=8, mode='max', restore_best_weights=True)

tuner.search(
    X_train_sm, y_train_sm,
    validation_data=(X_test_scaled, y_test),
    epochs=50,
    batch_size=64,
    callbacks=[early_stop_tuner],
    verbose=1
)

# Get best hyperparameters
best_hps = tuner.get_best_hyperparameters(num_trials=1)[0]
print('=== Best Hyperparameters ===')
print(f'  n_layers:      {best_hps.get("n_layers")}')
for i in range(best_hps.get('n_layers')):
    print(f'  Layer {i}: units={best_hps.get(f"units_{i}")}, dropout={best_hps.get(f"dropout_{i}")}')
print(f'  learning_rate: {best_hps.get("learning_rate"):.6f}')
```

## 8. Train Best ANN Architecture (from Tuner)

```python
# Build the best model from tuner
ann_best = tuner.hypermodel.build(best_hps)
ann_best.summary()

callbacks_final = [
    EarlyStopping(
        monitor='val_auc', patience=20, mode='max',
        restore_best_weights=True, verbose=1
    ),
    ReduceLROnPlateau(
        monitor='val_loss', factor=0.4, patience=8,
        min_lr=1e-7, verbose=1
    ),
    ModelCheckpoint(
        'best_ann_v3.keras', monitor='val_auc', mode='max',
        save_best_only=True, verbose=0
    )
]

history_best = ann_best.fit(
    X_train_sm, y_train_sm,
    batch_size=64,
    epochs=300,
    validation_data=(X_test_scaled, y_test),
    callbacks=callbacks_final,
    verbose=1
)
print('Best ANN training complete.')
```

## 9. 5-Fold Stratified Cross-Validation

> Ensures the reported metrics are stable and not lucky from a single split.
> SMOTE is applied **inside** each fold to prevent leakage.

```python
from sklearn.model_selection import StratifiedKFold

skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=SEED)
cv_aucs, cv_f1s, cv_recalls = [], [], []

X_all = np.vstack([X_train_sm, X_test_scaled])  # Use full dataset for CV
# Actually use original unsmoted data for fair CV
X_cv = scaler.fit_transform(X.values)
y_cv = y

print('Running 5-fold CV (SMOTE inside each fold)...')
for fold, (tr_idx, val_idx) in enumerate(skf.split(X_cv, y_cv), 1):
    X_tr, X_val = X_cv[tr_idx], X_cv[val_idx]
    y_tr, y_val = y_cv[tr_idx], y_cv[val_idx]

    # Apply SMOTE only to training fold
    X_tr_sm, y_tr_sm = SMOTE(random_state=SEED).fit_resample(X_tr, y_tr)

    # Build fresh model with best HPs
    fold_model = tuner.hypermodel.build(best_hps)
    fold_model.fit(
        X_tr_sm, y_tr_sm,
        validation_data=(X_val, y_val),
        epochs=100, batch_size=64, verbose=0,
        callbacks=[EarlyStopping(monitor='val_auc', patience=10, mode='max', restore_best_weights=True)]
    )

    y_prob_fold = fold_model.predict(X_val, verbose=0).flatten()
    best_t, _ = find_best_threshold(y_val, y_prob_fold)
    y_pred_fold = (y_prob_fold >= best_t).astype(int)

    fold_auc = roc_auc_score(y_val, y_prob_fold)
    fold_f1  = f1_score(y_val, y_pred_fold)
    fold_rec = recall_score(y_val, y_pred_fold)
    cv_aucs.append(fold_auc)
    cv_f1s.append(fold_f1)
    cv_recalls.append(fold_rec)
    print(f'  Fold {fold}: AUC={fold_auc:.4f} | F1={fold_f1:.4f} | Recall={fold_rec:.4f}')

print(f'\n=== 5-Fold CV Results ===')
print(f'AUC-ROC:  {np.mean(cv_aucs):.4f} +/- {np.std(cv_aucs):.4f}')
print(f'F1-Score: {np.mean(cv_f1s):.4f} +/- {np.std(cv_f1s):.4f}')
print(f'Recall:   {np.mean(cv_recalls):.4f} +/- {np.std(cv_recalls):.4f}')
```

```python
# CV results visualisation
cv_data = {'AUC-ROC': cv_aucs, 'F1-Score': cv_f1s, 'Recall': cv_recalls}
fig, axes = plt.subplots(1, 3, figsize=(15, 5))
fig.suptitle('5-Fold CV — Score Distribution', fontsize=14, fontweight='bold')
colors = ['#3498db', '#2ecc71', '#e74c3c']
for ax, (metric, scores), color in zip(axes, cv_data.items(), colors):
    ax.bar(range(1, 6), scores, color=color, alpha=0.85, edgecolor='white')
    ax.axhline(np.mean(scores), color='black', linestyle='--', linewidth=2, label=f'Mean={np.mean(scores):.4f}')
    ax.fill_between(range(0,7), np.mean(scores)-np.std(scores), np.mean(scores)+np.std(scores),
                    alpha=0.15, color=color)
    ax.set_title(metric, fontsize=12, fontweight='bold')
    ax.set_xlabel('Fold'); ax.set_ylim(0, 1.1); ax.legend(); ax.grid(alpha=0.3)
plt.tight_layout(); plt.show()
```

## 14. SHAP Feature Importance (Explainability)

> SHAP (SHapley Additive exPlanations) tells us which features most influence
> each individual prediction — making the model interpretable.

```python
# Use a background sample for SHAP (smaller for speed)
bg_sample = shap.sample(X_train_sm, 200, random_state=SEED)

# DeepExplainer works with Keras/TF models
explainer = shap.DeepExplainer(ann_best, bg_sample)
shap_values = explainer.shap_values(X_test_scaled[:500])

# shap_values shape: (500, n_features)
shap_arr = shap_values[0] if isinstance(shap_values, list) else shap_values

print(f'SHAP values computed: {shap_arr.shape}')
```

```python
# Global feature importance — mean absolute SHAP
mean_shap = np.abs(shap_arr).mean(axis=0)
shap_importance = pd.Series(mean_shap, index=feature_names).sort_values(ascending=True)

plt.figure(figsize=(10, 6))
colors_shap = ['#e74c3c' if v > shap_importance.median() else '#3498db'
               for v in shap_importance.values]
plt.barh(shap_importance.index, shap_importance.values, color=colors_shap, edgecolor='white')
plt.xlabel('Mean |SHAP Value|', fontsize=12)
plt.title('Global Feature Importance (SHAP)', fontsize=14, fontweight='bold')
plt.axvline(shap_importance.median(), color='gray', linestyle='--', linewidth=1, label='Median')
plt.legend(); plt.tight_layout(); plt.show()

print('Top 3 most influential features:')
print(shap_importance.sort_values(ascending=False).head(3).to_string())
```

```python
# SHAP summary bee-swarm plot
plt.figure(figsize=(10, 7))
shap.summary_plot(shap_arr, X_test_scaled[:500],
                  feature_names=feature_names, show=False)
plt.title('SHAP Summary Plot (Bee-Swarm)', fontsize=13, fontweight='bold')
plt.tight_layout(); plt.show()
```

```python
# SHAP waterfall for a single churner prediction
churner_idx = np.where(y_test == 1)[0][0]  # First churner in test set
shap_exp = shap.Explanation(
    values=shap_arr[churner_idx],
    base_values=explainer.expected_value[0] if isinstance(explainer.expected_value, list)
                else explainer.expected_value,
    data=X_test_scaled[churner_idx],
    feature_names=feature_names
)
plt.figure(figsize=(10, 6))
shap.plots.waterfall(shap_exp, show=False)
plt.title('SHAP Waterfall — Single Churner Explanation', fontsize=13, fontweight='bold')
plt.tight_layout(); plt.show()
```
